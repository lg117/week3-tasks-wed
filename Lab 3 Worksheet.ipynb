```python
from bs4 import BeautifulSoup
import re

with open('bongsoon.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

Check that the parsing was successful:

```python
soup.title
```

```python
meta = soup.find('meta', attrs={'name': 'description'})
description = meta['content']

print(description)
```

#### task a

```python
year = re.search(r'\d{4}', description)

print(year.group())
```

#### task b 

```python
words = re.findall(r'\b\w{6}\b', description)

print(words)
```

#### task c 

```python
print(words)
print(year.group())
```

#### ex 2

```python
synopsis = soup.find('div', class_='show-synopsis')
text = synopsis.find('p').get_text(strip=True)

print(text)
```

#### 2a

```python
vowel_words = re.findall(r'\b[aeiou]\w*\b', text, re.IGNORECASE)

print(vowel_words)
```

#### 2b

```python
no_ae = re.findall(r'\b[^ae\s]+\b', text, re.IGNORECASE)

print(no_ae)
```

#### ex 3

```python
link = soup.find('b', string='Viki').parent
href = link['href']

print(href)
```

#### 3a



```python
domain = re.search(r'www\.(.*?)%2F', href)

print(domain.group(1))
```

#### 3b 

```python
match = re.search(r'(https)%3A%2F%2Fwww\.(.*?)%2Ftv%2F(.*?)&', href)

print(match.group(1))
print(match.group(2))
print(match.group(3))
```

#### ex 5

```python
container = soup.find('div', class_='col-xs-8 col-sm-9 col-lg-9')
```

```python
container_text = container.get_text(" ", strip=True)

print(container_text)
```

```python
pattern = r'Ratings:\s*(?P<rating>\d+\.\d+)\s*/10 from\s*(?P<rating_users>[\d,]+).*?# of Watchers:\s*(?P<watchers>[\d,]+).*?Reviews:\s*(?P<review_users>[\d,]+)'
```

```python
match = re.search(pattern, container_text)

print(match.groupdict())
```
